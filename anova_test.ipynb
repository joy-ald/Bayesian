```python
# anova test of two sets of data
# read data from two csv files. Find the anova test variable and give the conclusion

# read data from csv
import numpy as np
from scipy import stats
# Example datasets
dataset1 = np.array([12, 15, 14, 10, 13, 16])
dataset2 = np.array([22, 25, 21, 23, 24, 26])
f_statistic, p_value = stats.f_oneway(dataset1, dataset2)

print("F-statistic:", f_statistic)
print("P-value:", p_value)
f_statistic, p_value = stats.f_oneway(dataset1, dataset2)

print("F-statistic:", f_statistic)
print("P-value:", p_value)
```

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Combine data into a DataFrame
data = pd.DataFrame({
    'value': np.concatenate([dataset1, dataset2]),
    'group': ['A'] * len(dataset1) + ['B'] * len(dataset2)
})

# ANOVA model
model = ols('value ~ group', data=data).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
```
